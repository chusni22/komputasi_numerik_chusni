```python
def eliminasi_gauss(A, b):
    n = len(b)

    # 1. Membuat Matriks Augmented [A | b]
    M = [[float(A[i][j]) for j in range(n)] + [float(b[i])] for i in range(n)]

    # 2. Proses Eliminasi Maju (Forward Elimination)
    for k in range(n):
        # Periksa pivot nol
        if M[k][k] == 0.0:
            raise ValueError("Terjadi pembagian dengan nol! Diperlukan teknik pivoting.")

        for i in range(k + 1, n):
            faktor = M[i][k] / M[k][k]
            for j in range(k, n + 1):
                M[i][j] -= faktor * M[k][j]

    # 3. Proses Substitusi Mundur (Back Substitution)
    x = [0.0 for _ in range(n)]
    for i in range(n - 1, -1, -1):
        jumlah = 0.0
        for j in range(i + 1, n):
            jumlah += M[i][j] * x[j]
        x[i] = (M[i][n] - jumlah) / M[i][i]

    return x

# --- Contoh Pengujian Sistem Persamaan Linier ---
# Persamaan:
# 2x + y + z = 1
# 4x + 3y + 2z = 2
# 2x + 10y + 2z = 6

A = [
    [2, 1, 1],
    [4, 3, 2],
    [2, 10, 2]
]

b = [1, 2, 6]

print("Matriks Koefisien (A):")
for row in A:
    print(row)

print("\nVektor Konstanta (b):", b)

# Menjalankan fungsi
solusi = eliminasi_gauss(A, b)

print("\n--- HASIL PENYELESAIAN ---")
for idx, val in enumerate(solusi):
    print(f"x{idx+1} = {val:.4f}")
```

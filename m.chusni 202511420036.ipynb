```python
def eliminasi_gauss(A, b):
  Mendefinisikan fungsi bernama eliminasi_gauss yang menerima dua parameter: matriks koefisien A dan vektor konstanta b.
    n = len(b)
    Menyimpan ukuran dimensi sistem persamaan ($n$) dengan menghitung jumlah elemen di dalam vektor b (atau jumlah baris matriks).


    M = [[float(A[i][j]) for j in range(n)] + [float(b[i])] for i in range(n)]
    Membuat matriks augmented dengan menggabungkan matriks koefisien A dan vektor konstanta b menjadi satu kesatuan tabel berukuran $n \times (n+1)$.Fungsi float(...) digunakan di sini agar semua angka diubah menjadi desimal, sehingga operasi pembagian nantinya menghasilkan angka pecahan yang akurat (bukan pembagian bulat).

    for k in range(n):
    Perulangan utama ($k$) untuk melangkah melalui setiap kolom pivot dari baris/kolom pertama hingga terakhir.

        if M[k][k] == 0.0:
            raise ValueError("Terjadi pembagian dengan nol! Diperlukan teknik pivoting.")
            Pengecekan pivot nol: Memeriksa apakah elemen di posisi diagonal utama (M[k][k]) bernilai nol. Jika iya, program akan berhenti dan mengeluarkan pesan error karena pembagian dengan nol tidak valid dalam matematika.

        for i in range(k + 1, n):
          Perulangan baris di bawah pivot ($i$) untuk mengakses semua baris yang berada di bawah baris pivot saat ini ($k$).
            faktor = M[i][k] / M[k][k]
            Menghitung faktor pengali: Mencari rasio antara elemen di bawah pivot dengan elemen pivot utamanya agar nanti bisa saling mengurangi hingga bernilai nol.
            for j in range(k, n + 1):
                M[i][j] -= faktor * M[k][j]
                Operasi Baris Elementer (OBE): Perulangan kolom ($j$) untuk mengurangkan setiap elemen pada baris $i$ dengan hasil kali antara faktor dan elemen pada baris pivot $k$. Langkah ini sukses mengubah elemen-elemen di bawah diagonal utama menjadi nol (membentuk matriks segitiga atas).


    x = [0.0 for _ in range(n)]
    Menyiapkan penampung hasil (x): Membuat sebuah list kosong berukuran $n$ yang diisi nilai awal 0.0 untuk menyimpan nilai akhir variabel nantinya ($x_1, x_2, \dots$).
    for i in range(n - 1, -1, -1):
      Perulangan mundur ($i$): Memulai proses perhitungan dari baris terbawah (persamaan terakhir yang variabelnya sudah tinggal satu) menuju baris paling atas.
        jumlah = 0.0
        for j in range(i + 1, n):
            jumlah += M[i][j] * x[j]
            Menghitung akumulasi nilai: Menjumlahkan hasil kali antara koefisien variabel lain yang nilainya sudah pernah dicari sebelumnya dengan nilai variabel tersebut (x[j]).
        x[i] = (M[i][n] - jumlah) / M[i][i]
        Mendapatkan nilai variabel: Menghitung nilai variabel pada baris ke-$i$ dengan cara mengurangi konstanta di ujung kanan matriks (M[i][n]) dengan jumlah, lalu membaginya dengan elemen diagonal utamanya (M[i][i]).

    return x
    Mengembalikan nilai akhir: Mengirimkan kembali hasil akhir berupa list nilai solusi variabel x kepada pemanggil fungsi agar bisa dicetak atau digunakan lebih lanjut.

A = [
    [2, 1, 1],
    [4, 3, 2],
    [2, 10, 2]
]

b = [1, 2, 6]

print("Matriks Koefisien (A):")
for row in A:
    print(row)

print("\nVektor Konstanta (b):", b)


solusi = eliminasi_gauss(A, b)

print("\n--- HASIL PENYELESAIAN ---")
for idx, val in enumerate(solusi):
    print(f"x{idx+1} = {val:.4f}")
```
